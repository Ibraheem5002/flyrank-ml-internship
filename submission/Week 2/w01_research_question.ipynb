Why:

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Question: Which of our top pages are starting to slip, and how do we decide which ones to fix first?

Unit of Analysis: A single page

Decision: Every end of week, a person reviews the pages which are falling off.

Action: The person sends the falling-off pages to the writers. They then update the pages.

Cost:
    False Postive (Flagging a good page): A writer wastes his time rewriting a page that was performing fine, wasting time and money.
    False Negative (Missing a bad page): By drawing no traffic, we have lose money.
